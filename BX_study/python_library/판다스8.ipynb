```python
import pandas as pd

# 1. 명사 리스트로 데이터프레임 생성
df_word = pd.DataFrame({'word': nouns})

# 2. 단어별 빈도수 계산 ('n' 컬럼 생성)
df_word = df_word.groupby('word', as_index=False) \
                 .agg(n=('word', 'count')) \
                 .sort_values('n', ascending=False)

# 3. 한 글자 단어 제외 (두 글자 이상만 추출)
df_word = df_word.query('word.str.len() >= 2')

# 4. 상위 결과 확인
df_word.head()
```

```python
# 1. 'word' 열을 데이터프레임의 인덱스로 설정
df_word.set_index('word')

# 결과 예시:
#          n
# word      
# 나라     19
# 일자리   19
# ...
# 수천      1

```

```python
# 2. 인덱스 설정 후 딕셔너리로 변환하여 특정 컬럼('n')의 값만 추출
dic_word = df_word.set_index('word').to_dict()['n']
dic_word
```
