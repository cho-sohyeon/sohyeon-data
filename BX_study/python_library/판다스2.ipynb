```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. 방어적 데이터 추출 함수
def get_text_safe(parent, selector, default="N/A"):
    """
    특정 선택자에서 텍스트를 추출하며, 요소가 없을 경우 기본값을 반환하여 
    NoneType 오류를 방지하는 함수입니다.
    """
    element = parent.select_one(selector)
    if element:
        return element.get_text(strip=True)
    return default

def scrape_yes24_bestsellers():
    # 요청할 URL (컨텍스트에서 제공된 베스트셀러 카테고리 주소)
    url = "https://www.yes24.com/product/category/bestseller?categoryNumber=001&pageNumber=1&pageSize=24"
    
    # 브라우저인 것처럼 속이기 위한 헤더 설정 (403 에러 방지)
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    }

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()  # 응답 코드가 200이 아니면 예외 발생
        soup = BeautifulSoup(response.text, "html.parser")

        # 핵심 선택자: 도서 목록 리스트 컨테이너
        book_list_container = soup.select_one("#yesBestList")
        
        if not book_list_container:
            print("도서 목록 컨테이너를 찾을 수 없습니다.")
            return pd.DataFrame()

        books_data = []

        # 각 도서 아이템(li) 순회
        # (일반적으로 yesBestList 하위의 li 요소를 찾습니다)
        for item in book_list_container.select("li"):
            # 데이터 수집 (제공된 핵심 선택자 활용)
            title = get_text_safe(item, "a.gd_name")             # 도서명
            publisher = get_text_safe(item, "span.authPub.info_pub") # 출판사/저자 정보
            price = get_text_safe(item, "em.yes_b")               # 가격
            reviews = get_text_safe(item, "em.txC_blue")          # 리뷰 수/인기도 관련

            # 제목이 없는 경우는 데이터가 없는 것으로 간주하고 스킵
            if title == "N/A":
                continue

            books_data.append({
                "도서명": title,
                "출판/저자": publisher,
                "가격": price,
                "리뷰수": reviews
            })

        # 데이터프레임 변환
        df = pd.DataFrame(books_data)
        return df

    except Exception as e:
        print(f"크롤링 중 오류가 발생했습니다: {e}")
        return pd.DataFrame()

# 메인 실행부
if __name__ == "__main__":
    df = scrape_yes24_bestsellers()
    
    # 마지막에 상위 3행만 출력
    if not df.empty:
        print("## 수집 결과 상위 3행 ##")
        print(df.head(3))
    else:
        print("수집된 데이터가 없습니다.")
```
