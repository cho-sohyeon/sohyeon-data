```python
#suv 차 종류에 대해서 합산 연비, 제조 회사 별로 분리, 내림차순, top 5만 본다. 
mpg.query("category == 'suv'") \
    .assign(total = (mpg["cty"] + mpg["hwy"]) / 2) \
    .groupby("manufacturer")  \
    .agg(mean_total = ("total", "mean")) \
    .sort_values("mean_total", ascending=False) 
```
