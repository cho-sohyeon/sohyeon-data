```python
"type" in df.columns 
```

```python
df["s_name"][9]
```

```python
"R" in df["s_name"][9]
```

```python
def store_type(x):
    type_name = ""
    if "R" in x:
        type_name ="R"
    elif "DT" in x:
        type_name = "DT"
    else: 
        type_name = "일반"
    return type_name 
```

```python
store_type(df["s_name"][103])
```

```python
df["s_name"][102]
```

```python
df["s_name"].apply(store_type)
```

```python
df["s_type"]  = df["s_name"].apply(store_type)
df.head(3)
```

```python
# 매장 타입 별 집계 
df.groupby("s_type").agg(type_count = ("s_type", "count"))
```

```python
df.groupby(["sido_name", "s_type"]).agg(type_count = ("s_type", "count"))
```

```python
# 2. lambda 식을 이용한 s_type 컬럼 생성
df['s_type'] = df['s_name'].apply(lambda x: 'R' if x.endswith('R') else ('DT' if x.endswith('DT') else '일반'))
```

```python
m = folium.Map(location=[36.5, 127.8], zoom_start=7)
```

```python
for _, row in df.iterrows():
    # 위도, 경도 : lat, lot 
    # 스토어 이름: s_name 
    # 스토어 주소: addr
    # 스토어 타입: s_type 
    print("위도경도: ", row["lat"], row["lot"])
    print("스토어 이름: ", row["s_name"])
    print("스토어 주소: ", row["addr"])
    print("스토어 타입: ", row["s_type"])
    popup_text = f"{row['s_name']},br.{row['addr']}<br>{row['s_type']}"
    popup = folium.Popup(popup_text, max_width=300)
    folium.Marker(
        location=[row["lat"], row["lot"]],
        popup = popup, 
        tooltip=row['s_name']
    ).add_to(m)
```

```python
m.save("result_map.html")
```
