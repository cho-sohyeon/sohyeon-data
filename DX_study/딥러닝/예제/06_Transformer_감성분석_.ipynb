```python
from transformers import pipeline

# 감정 분석 파이프라인 로드 (기본 모델 사용)
classifier = pipeline("sentiment-analysis")
```

```python
# 텍스트 예측 테스트
result = classifier("I love using Google Colab for deep learning!")
print(result)
```

```python
# 텍스트 예측 테스트
result = classifier("I'm sad that there aren't many deep learning classes left.")
print(result)
```

```python
 # 텍스트 예측 테스트
result = classifier("배송이 너무 늦어서 실망했어요.")
print(result)
```

기본 모델(distilbert-base-uncased-finetuned-sst-2-english)은 영어 전용이라 한글 문장에는 제대로 작동하지 않습니다.  
한국어를 지원하는 모델을 명시적으로 지정하면 됩니다.

```python
classifier = pipeline(
    "sentiment-analysis",
    model="matthewburke/korean_sentiment"
)
```

```python
# 텍스트 예측 테스트 - 한글
result = classifier("배송이 너무 늦어서 실망했어요.")
print(result)  # LABEL_0 = 부정, LABEL_1 = 긍정
```

```python
# 텍스트 예측 테스트 - 한글
result = classifier("공모전에서 입상을 했어요.")
print(result)  # LABEL_0 = 부정, LABEL_1 = 긍정
```

```python
# 한국어 지원 다국어 감성 분석 모델 지정
# 이 모델은 결과를 1~5점(1 star~5 stars)으로 반환합니다. 4~5는 긍정, 1~2는 부정, 3은 중립으로 해석하면 됩니다.
classifier = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)
```

```python
# 텍스트 예측 테스트 - 한글
result = classifier("배송이 너무 늦어서 실망했어요.")
print(result)
```
