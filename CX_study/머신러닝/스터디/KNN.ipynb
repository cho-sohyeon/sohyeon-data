```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

wine = load_wine()
X = wine.data
y = wine.target

k_values = [1, 2, 3, 4, 5]
test_sizes = [0.2, 0.3, 0.5]

results = []

for test_size in test_sizes:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42
    )
    for k in k_values:
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_train, y_train)
        pred = knn.predict(X_test)
        acc = accuracy_score(y_test, pred)
        results.append({
            'K': k,
            'test_size': test_size,
            '정확도': round(acc * 100, 1)
        })

df = pd.DataFrame(results)
df_pivot = df.pivot(index='K', columns='test_size', values='정확도')
df_pivot.columns = [f'test_size={c}' for c in df_pivot.columns]
print(df_pivot)
```

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

wine = load_wine()
X, y = wine.data, wine.target

k_values = range(1, 21)
test_sizes = [0.1, 0.15, 0.2, 0.25, 0.3, 0.4, 0.5]

results = []
for ts in test_sizes:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=ts, random_state=42)
    for k in k_values:
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_train, y_train)
        acc = accuracy_score(y_test, knn.predict(X_test))
        results.append({'K': k, 'test_size': ts, '정확도(%)': round(acc * 100, 1)})

df = pd.DataFrame(results)
best = df.loc[df['정확도(%)'].idxmax()]
print(df.pivot(index='K', columns='test_size', values='정확도(%)'))
print(f"\n최고 조합 → K={int(best.K)}, test_size={best.test_size}, 정확도={best['정확도(%)']}%")
```

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

wine = load_wine()
X, y = wine.data, wine.target

results = []
for k in range(1, 31):
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('knn', KNeighborsClassifier(n_neighbors=k))
    ])
    scores = cross_val_score(pipeline, X, y, cv=10)  # 10번 나눠서 검증
    results.append({
        'K': k,
        '평균 정확도(%)': round(scores.mean() * 100, 1),
        '표준편차(±)': round(scores.std() * 100, 1)
    })

df = pd.DataFrame(results).set_index('K')
best = df['평균 정확도(%)'].idxmax()
print(df)
print(f"\n최고 조합 → K={best}, 평균 정확도={df.loc[best, '평균 정확도(%)' ]}%, 표준편차=±{df.loc[best, '표준편차(±)']}%")
```

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

wine = load_wine()
X, y = wine.data, wine.target

k_values = range(1, 6)          # K=1~5로 좁히기
test_sizes = [0.2, 0.25, 0.3]
random_states = [0, 1, 42, 100, 200]  # 여러 번 섞어서 평균 내기

results = []
for ts in test_sizes:
    for k in k_values:
        accs = []
        for rs in random_states:
            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=ts, random_state=rs
            )
            scaler = StandardScaler()
            X_train = scaler.fit_transform(X_train)
            X_test = scaler.transform(X_test)

            knn = KNeighborsClassifier(n_neighbors=k)
            knn.fit(X_train, y_train)
            acc = accuracy_score(y_test, knn.predict(X_test))
            accs.append(acc * 100)

        results.append({
            'K': k,
            'test_size': ts,
            '평균 정확도(%)': round(sum(accs) / len(accs), 1),
            '최저(%)': round(min(accs), 1),
            '최고(%)': round(max(accs), 1)
        })

df = pd.DataFrame(results)
df_top = df[df['최고(%)'] < 100].sort_values(
    by=['평균 정확도(%)', 'test_size'],
    ascending=[False, False]
)
print(df_top.to_string(index=False))
print(f"\n최고 조합 → K={int(df_top.iloc[0]['K'])}, test_size={df_top.iloc[0]['test_size']}, 평균 정확도={df_top.iloc[0]['평균 정확도(%)']}%")
```
