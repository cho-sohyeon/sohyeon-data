```python
# 22번

import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import numpy as np

# (1) bmi_500.csv data 불러오기
bmi = pd.read_csv("/content/drive/MyDrive/colab notebooks/bmi_500.csv")
data = ["label"]

# (2) bmi data를 문제와 답으로 분리
X = data(["question"].axis=1)
y = data["answer"] 

# (3) LogisticRegression 분류 모델 생성
model = LogisticRegression(max_iter=100)
                        
         
# (4) 5-fold 교차검증 수행
scores = cross_val_score(model, X, y, cv=5, scoring="accuracy")


# (5) 교차검증 결과와 평균 정확도 출력
print("교차 검증 점수:", scores)
print("평균 정확도:", scores.mean())

```
